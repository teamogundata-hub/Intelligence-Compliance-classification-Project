```python
# ============================================================
# CBN PDF Loader and Text Extractor — Google Colab
# ============================================================

!pip -q install pdfplumber requests

import io
import json
import logging
from datetime import datetime
from pathlib import Path

import pdfplumber
from google.colab import files


logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)
logger = logging.getLogger("cbn_scraper")


class CBNCircularScraper:
    """Upload a PDF, extract its text, and save the result as JSON."""

    def __init__(self, output_dir="/content/data/raw_corpus"):
        self.output_dir = Path(output_dir)
        self.output_dir.mkdir(parents=True, exist_ok=True)

    def upload_pdf(self):
        """Open a file picker and return uploaded PDF filename and bytes."""
        print("Click 'Choose Files' and select your PDF.")

        uploaded = files.upload()

        if not uploaded:
            raise ValueError("No file was uploaded.")

        filename = next(iter(uploaded))
        pdf_content = uploaded[filename]

        if not filename.lower().endswith(".pdf"):
            raise ValueError("Please upload a PDF file ending in .pdf.")

        if not pdf_content.startswith(b"%PDF"):
            raise ValueError("The selected file is not a valid PDF.")

        logger.info(f"Uploaded: {filename} ({len(pdf_content):,} bytes)")
        return filename, pdf_content

    def extract_text(self, pdf_content):
        """Extract text from PDF bytes."""
        try:
            with pdfplumber.open(io.BytesIO(pdf_content)) as pdf:
                pages = [
                    page.extract_text() or ""
                    for page in pdf.pages
                ]

            text = "\n\n".join(pages)

            logger.info(
                f"Extracted {len(text):,} characters "
                f"from {len(pages)} pages."
            )
            return text

        except Exception as error:
            logger.error(f"Text extraction failed: {error}")
            return ""

    def save_document(self, title, filename, text):
        """Save extracted text and metadata to JSON."""
        output_file = self.output_dir / "cbn_document.json"

        document = {
            "title": title,
            "original_filename": filename,
            "source": "Manual PDF upload",
            "collected_at": datetime.utcnow().isoformat(),
            "text": text,
        }

        with open(output_file, "w", encoding="utf-8") as file:
            json.dump(document, file, indent=2, ensure_ascii=False)

        logger.info(f"Saved JSON to: {output_file}")
        return output_file


# ============================================================
# RUN: Upload, extract, preview, and save
# ============================================================

scraper = CBNCircularScraper()

filename, pdf_content = scraper.upload_pdf()

text = scraper.extract_text(pdf_content)

print("\n" + "=" * 80)
print(f"FILE: {filename}")
print(f"CHARACTERS EXTRACTED: {len(text):,}")
print("=" * 80)
print(text[:3000])

output_file = scraper.save_document(
    title=Path(filename).stem,
    filename=filename,
    text=text
)

print("\n" + "=" * 80)
print(f"Saved JSON output: {output_file}")
```
